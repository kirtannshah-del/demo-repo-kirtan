```python
import random

print("PATTERNIO")

print("\n a hidden rule is generating numbers")
print("\n your job : guess the next output")

score=0

rules=[lambda x:x*5-5,lambda x:x*2+2,lambda x:x+5,lambda x:x*3-1]

rule=random.choice(rules)
current=random.randint(1,10)

rounds=5

for i in range(rounds):

    print(f"ROUND{i+1}")

    print(f"Given number:{current}")

    answer=rule(current)

    guess = input("Your guess of NEXT number: ")

    if guess.isdigit() and int(guess) == answer:
        print(" Correct!")
        score += 1
    else:
        print(f"Wrong! Correct was {answer}")


    current = answer

print("GAME OVER")

print(f"Final Score: {score}/{rounds}")

if score == rounds:
    print(" PERFECT DETECTIVE! You are a crazy genius.")
elif score >= 3:
    print( "Good logic skills!")
else:
    print(" Pattern was too hard this time.")
```
